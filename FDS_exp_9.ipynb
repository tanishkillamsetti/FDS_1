```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('iris.csv')

print(df.head())
print(df.info())
print(df['species'].value_counts())
```

```python
#histogram
plt.hist(df['sepal_length'], bins=10, edgecolor='black')
plt.title('Histogram of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.show()
```

```python
#scatter plot
sns.scatterplot(
    data=df,
    x='sepal_length',
    y='petal_length',
    hue='species'
)
plt.title('Sepal Length vs Petal Length')
plt.show()
```

```python
#line plot
plt.plot(df.index, df['sepal_length'])
plt.title('Sepal Length Across Observations')
plt.xlabel('Observation Number')
plt.ylabel('Sepal Length')
plt.show()
```

```python
#barplot
sns.barplot(data=df, x='species', y='petal_length', errorbar=None)
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Length')
plt.show()
```

```python
#count plot
sns.countplot(data=df, x='species')
plt.title('Count of Each Iris Species')
plt.xlabel('Species')
plt.ylabel('Number of Flowers')
plt.show()
```

```python
#boxplot
sns.boxplot(data=df, x='species', y='sepal_width')
plt.title('Sepal Width by Species')
plt.show()
```

```python
#heat map
correlation = df.select_dtypes(include='number').corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Iris Features')
plt.show()
```

```python
#pie chart
species_count = df['species'].value_counts()

plt.pie(
    species_count,
    labels=species_count.index,
    autopct='%1.1f%%',
    startangle=90,
    explode=[0.05, 0.05, 0.05]
)

plt.title('Percentage Distribution of Iris Species')
plt.axis('equal')
plt.show()
```
