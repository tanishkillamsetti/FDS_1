```python
import pandas as pd
emp1 = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105],
    "Name": ["Mithali", "Samiksha", "Jyothish", "Tanish", "Suhani"],
    "Dept": ["IT", "HR", "Finance", "IT", "Marketing"]
})

print("\n Employe :- 1")
print(emp1)
emp2 = pd.DataFrame({
    "Emp_ID": [106, 107, 108, 109, 110],
    "Name": ["Praneeth", "Ashish", "Lohit", "Tharun", "neeraj"],
    "Dept": ["HR", "IT", "Finance", "Marketing", "IT"]
})

print("\n Employe:- 2")
print(emp2)
salary = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105,
               106, 107, 108, 109, 110],
    "Salary": [45000, 40000, 55000, 50000, 42000,
               48000, 46000, 58000, 44000, 52000]
})

print("\n salary")
print(salary)
```

```python
#concatination

emp = pd.concat([emp1, emp2], ignore_index=True)
print(emp)
#concatination using columns
df1 = pd.DataFrame({
    "Name": ["Tanish", "Praneeth", "Ashish"],
    "Age": [20, 21, 20]
})

df2 = pd.DataFrame({
    "Marks": [85, 78, 92],
    "Grade": ["A", "B", "A"]
})

concat_columns = pd.concat([df1, df2], axis=1)
print(concat_columns)
```

```python

# INNER MERGE
# Only matching Emp_ID values

inner_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="inner"
)

print("\n inner merge")
print(inner_merge)
# LEFT MERGE
# All records from left DataFrame

left_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="left"
)

print("\n left merge ")
print(left_merge)
#  RIGHT MERGE
# All records from right DataFrame

right_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="right"
)

print("\n Right merge")
print(right_merge)
# OUTER MERGE
# All records from both DataFrames

outer_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="outer"
)

print("\nOUTER MERGE ")
print(outer_merge)
```

```python

# JOIN
# Join DataFrames using INDEX

emp_details = pd.DataFrame({
    "Name": ["Tanish", "Praneeth", "Ashish"],
    "Dept": ["IT", "HR", "Finance"]
}, index=[101, 102, 103])


emp_salary = pd.DataFrame({
    "Salary": [50000, 48000, 46000]
}, index=[101, 102, 103])


join_result = emp_details.join(emp_salary)

print("\n JOIn ")
print(join_result)

# LEFT JOIN

left_join = emp_details.join(
    emp_salary,
    how="left"
)

print("\n LEFT JOIN ")
print(left_join)
# RIGHT JOIN

right_join = emp_details.join(
    emp_salary,
    how="right"
)

print("\n RIGHT JOIN ")
print(right_join)
# OUTER JOIN
outer_join = emp_details.join(
    emp_salary,
    how="outer"
)

print("\n  OUTER JOIN ")
print(outer_join)
```

```python
#FINAL EMPLOYEE DATA WITH SALARY

final_df = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="inner"
)

print("\n FINAL EMPLOYEE DATA ")
print(final_df)
```
