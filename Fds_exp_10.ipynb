```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Automobile_data.csv')

numeric_columns = [
    'price', 'horsepower', 'peak-rpm',
    'bore', 'stroke', 'normalized-losses'
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

print(df.head())
print(df.info())
```

```python
#histogram
plt.hist(df['engine-size'].dropna(), bins=10, edgecolor='black')
plt.title('Histogram of Engine Size')
plt.xlabel('Engine Size')
plt.ylabel('Frequency')
plt.show()
```

```python
#scatter plot
sns.scatterplot(data=df, x='horsepower', y='price')
plt.title('Horsepower vs Automobile Price')
plt.xlabel('Horsepower')
plt.ylabel('Price')
plt.show()
```

```python
#line plot
plt.plot(df.index, df['price'])
plt.title('Automobile Price Across Observations')
plt.xlabel('Observation Number')
plt.ylabel('Price')
plt.show()
```

```python
#barplot
sns.barplot(
    data=df,
    x='body-style',
    y='price',
    errorbar=None
)
plt.title('Average Price by Body Style')
plt.xlabel('Body Style')
plt.ylabel('Average Price')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
#countplot
sns.countplot(data=df, x='fuel-type')
plt.title('Number of Cars by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Number of Cars')
plt.show()
```

```python
#box plot
sns.boxplot(data=df, x='body-style', y='price')
plt.title('Price Distribution by Body Style')
plt.xlabel('Body Style')
plt.ylabel('Price')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
#heatmap
numeric_df = df.select_dtypes(include='number')

correlation = numeric_df.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title('Correlation Heatmap of Automobile Features')
plt.tight_layout()
plt.show()
```

```python
#pie chart
body_counts = df['body-style'].value_counts()

plt.pie(
    body_counts,
    labels=body_counts.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Percentage Distribution of Automobile Body Styles')
plt.axis('equal')
plt.show()
```
