```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
# Select Age column and do operations
age = df["Age"].dropna()
print("\nRange:")
print(age.max() - age.min())
print("\nVariance:")
print(age.var())
print("\nStandard Deviation:")
print(age.std())
```

```python
print("\nQuartiles:")
print(age.quantile([0.25, 0.50, 0.75]))
Q1 = age.quantile(0.25)
Q3 = age.quantile(0.75)
IQR = Q3 - Q1
print("\nIQR:")
print(IQR)
```

```python
# Histogram
plt.hist(age, bins=10)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

```python
# Box plot
plt.boxplot(age)
plt.ylabel("Age")
plt.title("Age Box Plot")
plt.show()
```
