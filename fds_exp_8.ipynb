```python

import pandas as pd
#CREATE DATAFRAMES

students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Tanish", "Praneeth", "Ashish",
             "Lohit", "Tharun", "Chaitanya"],
    "Department": ["CSE", "CSE", "CSE",
                   "ECE", "ECE", "ECE"],
    "Year": [3, 3, 3, 3, 3, 3]
})

marks = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106],
    "Python": [85, 78, 90, 75, 88, 82],
    "Java": [80, 82, 85, 70, 90, 78],
    "DBMS": [88, 76, 92, 80, 85, 86]
})

print("STUDENT DATAFRAME")
print(students)

print("\nMARKS DATAFRAME")
print(marks)
```

```python
# HIERARCHICAL INDEXING

hierarchical = students.set_index(["Department", "Student_ID"])

print(hierarchical)

# Select students belonging to CSE
print("\nCSE STUDENTS:")
print(hierarchical.loc["CSE"])

# Select one particular student
print("\nSTUDENT 101:")
print(hierarchical.loc[("CSE", 101)])
```

```python
# COMBINING DATAFRAMES USING CONCAT
# Split into two parts
df1 = students.iloc[:3]
df2 = students.iloc[3:]

# Combine rows
combined_rows = pd.concat([df1, df2], ignore_index=True)

print("\nCOMBINE ROWS:")
print(combined_rows)

# Combine columns
combined_columns = pd.concat(
    [students, marks.drop(columns=["Student_ID"])],
    axis=1
)

print("\nCOMBINE COLUMNS:")
print(combined_columns)
```

```python
# MERGING DATAFRAMES
# Inner merge
inner_merge = pd.merge(
    students, marks,
    on="Student_ID",
    how="inner"
)

print("\nINNER MERGE:")
print(inner_merge)

# Left merge
left_merge = pd.merge(
    students, marks,
    on="Student_ID",
    how="left"
)

print("\nLEFT MERGE:")
print(left_merge)

# Right merge
right_merge = pd.merge(
    students, marks,
    on="Student_ID",
    how="right"
)

print("\nRIGHT MERGE:")
print(right_merge)

# Outer merge
outer_merge = pd.merge(
    students, marks,
    on="Student_ID",
    how="outer"
)

print("\nOUTER MERGE:")
print(outer_merge)
```

```python
# RESHAPING USING MELT
# WIDE TO LONG
long_data = pd.melt(
    marks,
    id_vars=["Student_ID"],
    value_vars=["Python", "Java", "DBMS"],
    var_name="Subject",
    value_name="Marks"
)

print(long_data)
```

```python
# PIVOTING USING PIVOT
# LONG TO WIDE
pivot_data = long_data.pivot(
    index="Student_ID",
    columns="Subject",
    values="Marks"
)

print(pivot_data)
```

```python
# PIVOT TABLE
# CALCULATE AVERAGE MARKS
merged_data = pd.merge(
    students, marks,
    on="Student_ID"
)
pivot_table_data = pd.pivot_table(
    merged_data,
    values=["Python", "Java", "DBMS"],
    index="Department",
    aggfunc="mean"
)
print(pivot_table_data)
```

```python
# STACK
# CONVERT COLUMNS INTO INDEX LEVELS

stack_data = pivot_data.stack()

print(stack_data)
```

```python
# UNSTACK
# CONVERT INDEX LEVEL BACK INTO COLUMNS
unstack_data = stack_data.unstack()

print(unstack_data)
```
