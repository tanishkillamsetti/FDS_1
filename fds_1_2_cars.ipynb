```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("Automobile_data.csv")
```

```python
sns.countplot(x='fuel-type', data=df)
plt.title("fuel types")
plt.show()
```

```python
sns.countplot(x='num-of-doors', data=df)
plt.title("Door count")
plt.show()
```

```python
sns.countplot(x='body-style', data=df)
plt.title("type of veichle ")
plt.show()
```

```python
drive_wheels_counts = df['drive-wheels'].value_counts()
plt.pie(drive_wheels_counts, labels=drive_wheels_counts.index, autopct='%1.1f%%', startangle=90)
plt.title("Type of Drive")
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
sns.countplot(x='engine-type', data=df)
plt.title("type of the engine ")
plt.show()
```

```python
sns.histplot(x='engine-size', data=df, kde=True)
plt.title("size chart ")
plt.show()
```
